```python
import pandas as pd
import numpy as np

spotify_cols = ['track_id', 'artists', 'track_name', 'popularity', 'duration_ms', 'danceability', 'valence']
df_spotify = pd.read_csv('../data/spotify.csv', usecols=spotify_cols)

print("--- SPOTIFY PROFILING ---")
print(f"Total filas: {len(df_spotify)}")
print("\n1. Conteo de Nulos por columna:")
print(df_spotify.isnull().sum())

print("\n2. Descripción estadística de numéricos (popularity, duration_ms, danceability, valence):")
print(df_spotify.describe().round(2))

print("\n3. Muestra del formato de la columna 'artists' (primeras 3 filas):")
print(df_spotify['artists'].head(3).tolist())
```

```python
# Cargar el dataset de Grammys

grammy_cols = ['year', 'category', 'nominee', 'artist', 'winner']
df_grammys = pd.read_csv('../data/grammy.csv', usecols=grammy_cols)

print("--- GRAMMY PROFILING ---")
print(f"Total filas: {len(df_grammys)}")
print("\n1. Conteo de Nulos por columna:")
print(df_grammys.isnull().sum())

print("\n2. Valores únicos en 'winner' (¿solo hay True?):")
print(df_grammys['winner'].value_counts(dropna=False))

print("\n3. Análisis de duplicados en 'artist' (Top 5 artistas con más nominaciones/premios):")
print(df_grammys['artist'].value_counts().head(5))

print("\n4. Muestra del formato de la columna 'artist' (primeras 3 filas):")
print(df_grammys['artist'].head(3).tolist())
```

```python
# Muestra de un artista en Spotify (ej. U2)
print("Formato en Spotify:")
print(df_spotify[df_spotify['artists'].str.contains('U2', na=False, case=False)]['artists'].head(2).tolist())

# Muestra del mismo artista en Grammys
print("\nFormato en Grammys:")
print(df_grammys[df_grammys['artist'].str.contains('U2', na=False, case=False)]['artist'].head(2).tolist())
```
